## Tutorial 1. Implementing logistic regression using a generated dataset

We go through 7 steps in logistic regression.

```python
# Step 1. Import the required libraries

# import pandas as pd
# import seaborn as sns #for statistics ploting

import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix

# from sklearn.metrics import classification_report

%matplotlib inline
```

### Use make_classification from sklearn.datasets to generate a dataset for classification

Refer to https://scikit-learn.org/stable/modules/generated/sklearn.datasets.make_classification.html for further information.

The dataset generated basically consists of $X$, the feature array of the data points, and $y$, the classification label array. 

Note that both $X$ and $y$ are ndarrays, i.e. NumPy arrays, which can represent both matrices (n-D) and vectors (1-D). 

https://numpy.org/doc/stable/user/absolute_beginners.html

$X$ is two-dimensional array, which represents a matrix, with each row being a one dimensional array itself which represents $n$ features of a data point in an $n$ dimensional feature space; the number of rows is the number of data points. $y$ is one-dimensional array, which represents a vector, with each element representing a class label, i.e. the ground truth of the corresponding data point.

We can simply print $X$ and $y$ to see what they are like. We can see that their representations are different from arrays in Python.

```python
# Step 2. Generate the dataset

X, y = make_classification(
    n_samples=100, # 100 data ponts
    n_features=1, # using 1 feature as an example - try a dataset with multiple features, e.g. 2 features
    n_classes=2, # binary classification
    n_clusters_per_class=1,
    flip_y=0.03,
    n_informative=1,
    n_redundant=0,
    n_repeated=0)

# print the first few rows of X
print(X[0:5,:])
```

As we can see, $X$ is a two dimensional ndarray in which each row is an array itself consisting of n features.

```python
# print the first few elements of y
print(y[0:5])
```

As we can see, $y$ is a one dimensional ndarray in which each element is the class label for a sample.

```python
print(X.shape) # use the shape function to show the numbers of rows and columns in two arrays
print(y.shape)
```

As we can see, $X$ has 100 rows in total for the 100 data points and 1 column for 1 feature only, $y$ has 100 elements for the 100 class labels.

### Display the sample data points

We can simply plot all the data points with the two ndarrays. For further information about Pyplot, refer to:

https://matplotlib.org/stable/tutorials/introductory/pyplot.html#sphx-glr-tutorials-introductory-pyplot-py

```python
# Step 3. Visualise the dataset
plt.scatter(X, y, c=y, cmap='rainbow')
plt.title("Logistic Regression Illustration")
plt.xlabel('X')
plt.ylabel('y')
plt.show
```

### Train and Test the logistic model

We first split the dataset into two datasets for training and testing. 

We then train the model with the trainig dataset and test the trained model with the testing dataset.

```python
# Step 4. Split the dataset into the training and testing datasets
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=1) # the default split is 75:25
# X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=1) # or choose our own split rate
```

```python
# Find out the numbers of rows and columns in an ndarray
print(X_train.shape) # use the shape function to show shape (sizes of rows and columns) of an ndarray
print(y_train.shape)
```

```python
# Step 5. Fit the logistic regression model
log_reg = LogisticRegression() # create an instance of the model
log_reg.fit(X_train, y_train) # train the model
```

### Use the predict function with the logistic model to make predictions

Once the model is trained, the predict function can be used to make predictions on new data points, which will produce as output the classifications of the data points.

The predict outputs the predictions only rather than the probabilities produced by the sigmoid function. 

```python
# Step 6. Make prediction using the trained model with the test dataset
y_pred = log_reg.predict(X_test)
```

For example, we can print first two data points in the testing datasets and the predictions on them as show below.

```python
print(X_test[0:2])
print(y_pred[0:2])
```

### Use confusion_matrix function to calculate and display the confusion matrix

The function takes the ground truth and the predictions as input and produces the confusion matrix as output, which is a 2-d ndarray.

```python
# Step 7. Display the confusion matrix 
print('Confusion Matrix:')
print(confusion_matrix(y_test,y_pred))
```

### Display the trained model along with the data points

We can display the model on all the data points in the dataset only. 

Alternatively if we want to see a more complete display of the model, we can first generate a full range of data points (instead of the data points in the dataset only), i.e., a full range of $X$ and their predictions, $y$; We can then display the full range of connected data points.

```python
# Step 8. Display the trained model
plt.scatter(X, y, c=y, cmap='rainbow', label='Data Points')
plt.title("Logistic Regression Illustration")
y_hat = log_reg.predict(X)
plt.scatter(X, y_hat, c='g', label='New Model') 

plt.legend(loc='best')

plt.show

print('Confusion Matrix:')
print(confusion_matrix(y,y_hat))
```

## Tutorial 2. Fit a logistic model on a generated dataset, and generate and plot the decision boundary 

### Generate and plot a linearly separatable binary dataset

Note that in the dataset generated, there are two elements, one (samples[0]) is $X$, and another (samples[1]) is $y$.

As the two classes may have overlap, we choose a dataset in which the two classes have not overlap hence separable.

```python
separable = False
while not separable:
    samples = make_classification(n_samples=100, n_features=2, n_redundant=0, n_informative=1, n_clusters_per_class=1, flip_y=-1)
    red = samples[0][samples[1] == 0] # red contains those data points that have y==0
    blue = samples[0][samples[1] == 1] # blue contains those data points that have y==1
    # check whether red and blue are separable
    separable = any([red[:, k].max() < blue[:, k].min() or red[:, k].min() > blue[:, k].max() for k in range(2)])
    X, y = samples
plt.plot(red[:, 0], red[:, 1], 'r.', marker="o")
plt.plot(blue[:, 0], blue[:, 1], 'b.', marker="x")
plt.show()
```

### Fit a logistic model to the dataset

```python
# Import and fit the logistic regression model
from sklearn import linear_model
log_reg = linear_model.LogisticRegression(C=1e5) # C is the inverse of regularization strength
log_reg.fit(X, y)

# Print the intercept and co-efficients with the logistic regression model
print('Co-efficients:', log_reg.coef_)
print('Intercept:', log_reg.intercept_)
```

**Note that log_reg.coef_ gets co-efficients for $x_1$ and $x_2$, $\theta_1$ and $\theta_2$ while log_reg.intercept_ get the intercept for $x_0$ (with a dummy value of 1, $\theta_0$, which is also referred to as bias)**

### Generate and plot the decision boundary with the dataset

Note that the decision boundary is represented by this equation: $\theta_0 + \theta_1x_1 + \theta_2x_2 = 0$. So we have $\theta_2x_2=-\theta_0 - \theta_1x_1$. Furthermore dividing both sides by $\theta_2$, we have $x_2=(-\theta_1/\theta_2)x_1-\theta_0/\theta_2$.

Note how $x_2$ is calculated in the cell below.

```python
import numpy as np

#w = log_reg.coef_[0]
#a = -w[0,0] / w[0,1]

a = -log_reg.coef_[0, 0] / log_reg.coef_[0, 1] 

# only a small range of x1 is generated so the decision boundary between samples is shown
# for a bigger range of x1, the range of x2 will much bigger

x1 = np.linspace(-0.75, 0.3) # the scale needs to be adjusted with reference to the plot for the samples

# x1 = np.linspace(-2.0, 2.0) # can uncomment this line and comment out the previous line to see what happens

x2 = a * x1 - (log_reg.intercept_[0]) / log_reg.coef_[0, 1]

plt.plot(red[:, 0], red[:, 1], 'r.', marker="o", label='0s')
plt.plot(blue[:, 0], blue[:, 1], 'b.', marker="x", label='1s')
plt.plot(x1, x2, 'g', linewidth=2, label='DB') 
plt.legend(loc='best')
plt.xlabel('$x_1$', fontsize='large')
plt.ylabel('$x_2$', fontsize='large')
plt.show()
```

## Exercise 3. Apply logistic regression to the iris dataset

The Iris dataset consists of 3 different types of irises: Setosa, Versicolour and Virginica, stored in a 150x4 ndarray.

The rows are the samples and the columns are the features: Sepal Length, Sepal Width, Petal Length and Petal Width.

The dataset is one of the toy datasets in sklearn, which can be loaded:

https://scikit-learn.org/stable/datasets/toy_dataset.html

See https://en.wikipedia.org/wiki/Iris_flower_data_set for more information on this dataset.

It turns out that the first 50 samples are for Setosa, the next 50 for Versicolour and the last 50 for Virginica. In this exercise, we'll use the first 100 samples only so the dataset is for binary classification: Setosa and Versicolour. We'll use the first two features only (Sepal Length and Sepal Width) so we have a 2 dimensional feature space for simplicity of visualisation. 

```python
import matplotlib.pyplot as plt
from sklearn import datasets

# import some data to play with
iris = datasets.load_iris()

print(iris.feature_names)
print(iris.target_names)

# we only take the first 100 samples 
X = iris.data[0:100, :2] # the first two features
y = iris.target[0:100] # for the first two classes only

print(X.shape)
print(y.shape)

x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5

plt.figure(2, figsize=(8, 6))

# Plot the training points
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='rainbow')
plt.xlabel("Sepal length (cm)")
plt.ylabel("Sepal width (cm)")

plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
#plt.xticks(())
#plt.yticks(())
plt.show()
```
