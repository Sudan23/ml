### Data structures

There are two primary data structures in Pandas, the DataFrame to represent tabular data with rows and named columns, and the Series to represent a single column of data. Hence a DataFrame is made up of one or more Series/column.

A dataframe or series can then be manipulated for data preparation. Quite often, the datasets we have are in tables. In these cases, using dataframes to represent them is convenient.

```python
# create a DataFrame with our own data

import pandas as pd # pandas is imported as pd by convention
books = pd.Series(['Harry Potter and The Philosophers Stone','Artificial Intelligence: A Modern Approach'])
book_class = pd.Series(['Fiction', 'Non-Fiction'])
authors = pd.Series(['JK Rowling', 'Russel and Norvig'])
sold = pd.Series([120000000, 10000]) # note these may not be true sales figures

# a DataFrame is created, which consists of the above four series together with their column headings
book_frame = pd.DataFrame({'Book Title': books, 'Book Classification': book_class,
'Author Name': authors, 'Sales': sold})
book_frame # the value of the last expression in a cell is displayed in a well-formatted table
```

```python
print(book_frame) # use a simple print function will display the value of the expression too
```

### Read data from a file

However, we typically read the data from a csv file using the function read_csv(). The file can be either in a local directory or from a remote site over HTTP.

### Accessing a data frame

You can access a DataFrame much the same as normal Python data structures. <i>Note that the first range is for columns and the second for rows.<i>

```python
print(book_frame['Book Title']) # access the named column
print(book_frame['Book Title'][1]) # access the indexed row of the named column
print(book_frame['Book Title'][0:2]) # access 2 rows of the named column starting from the indexed row
```

You can access slices of rows. The first range is for rows and the second for columns.

```python
# Run each of the statements in the cell (by commenting out the others) and see what's printed.
print(book_frame.iloc[0]) # print row 0
print(book_frame.iloc[1]['Book Title']) # print row 1 in the named colums
print(book_frame.iloc[0:1])# print 1 row starting from row 0
```

### Manipulation
You can manipulate the data within your dataframes.

```python
# Run each block of the statements in the cell (by commenting out the others) and see what's printed.

#book_frame['Sales'] = book_frame['Sales'] * 10 # increase the sales of books by a factor of 10
#print(book_frame)

import numpy as np
# a new series is created (with the natural (base e) logorithm of each sale figure
log_sales = np.log(book_frame['Sales']) 
print(log_sales)
book_frame['Log_Sales'] = log_sales # a new named colum is added to a dataframe
print(book_frame)
```

### Figures and axes

Two important data structures used in matplotlib are figures and axes. Figures correspond to physical figures, while axes corresponds to regions or sections of a figure. Typically, you first create a figure and add axes to it. We can plot different things in a figure or axe once it has been created.

```python
%lsmagic # display available line and cell magics
```

**The %matplotlib inline is a jupyter notebook specific command that allows us to see the plots in the notebook itself.**

```python
# the following command get figures displayed inline
%matplotlib inline 
import matplotlib.pyplot as plt
fig = plt.subplots(1) # create a figure with 2 axes for two generated plots
fig, (ax1, ax2) = plt.subplots(2) # create a figure with 2 axes for two generated plots
fig, (ax1, ax2) = plt.subplots(2, sharex=True, sharey=True) # same as above with x and y axes shared
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3)
sigma=3 # the standard deviation in the Gaussian distribution 
mu = 2 # the mean
n = sigma * np.random.randn(500,1) + mu
ax1.plot(n) # plot a line in this axe 
# add some noise to n for visualisation purposes
mu_noise, sigma_noise = 1, .7
noise = sigma_noise * np.random.randn(500,1) + mu_noise
y = n * noise
ax2.scatter(n, y) # plot scatters in this axe
ax3.hist(n)
# annoyingly we sometimes need this to display plots with better spacing
plt.tight_layout()
```

```python
"""
Simple demo of a scatter plot
"""
import numpy as np
import matplotlib.pyplot as plt

N = 50
x = np.random.rand(N) # creates a series of N real numbers in the range of 0 to 1 - note no column, just a series
print(x)
y = np.random.rand(N)
print(y)
colours = np.random.rand(N) # N randomly generated colours
area = np.pi * (15 * (np.random.rand(N)))**2 
plt.scatter(x, y, s=area, c=colours, alpha=0.5) # each spot is the size of randomly generated area
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

X = 2 * np.random.rand(10, 1) # create a 2-d array of 10 rows and 1 column
y = 4 + 3 * X + np.random.randn(10, 1)
y = X + np.random.randn(10, 1)
print(X)
print(y)

colours = np.random.rand(10) # 10 randomly generated colours
plt.scatter(X, y, c=colours) # X and y are both 2-d arrays but the scatters are the same as the ones by two series
plt.show()
```
